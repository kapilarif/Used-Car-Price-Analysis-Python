## 4. NumPy Analysis

In this section, NumPy is used to perform numerical analysis on the car selling prices.

The following operations are performed:

- Convert selling prices into a NumPy array.
- Calculate the average selling price.
- Find the maximum selling price.
- Find the minimum selling price.
- Calculate the standard deviation.

```python
## Numpy Calculations

print("----------Numpy Calculations----------")

import numpy as np

print("----------Mean-------------\n")
print(np.mean(df["Selling_Price"]),"\n")

print("---------Maximum---------\n")
print(np.max(df["Selling_Price"]),"\n")

print("---------Minimum----------\n")
print(np.min(df["Selling_Price"]),"\n")

print("----------Standard Deviation-----------\n")
print(np.std(df["Selling_Price"]),"\n")


```

## 5. Statistical Analysis

In this section, statistical methods are used to understand the distribution and variation of used car selling prices.

The following statistical measures are calculated:

- Mean – Average selling price.
- Median – Middle value of selling prices.
- Mode – Most frequently occurring selling price.
- Variance – Measures the spread of prices.
- Standard Deviation – Measures how much prices vary from the mean.

```python
## Statistics Calculations

print("----------Statistics Calculations--------------\n")

import statistics as st
print("----------Mean--------\n")
print(st.mean(df["Selling_Price"]),"\n")

print("----------Median--------\n")
print(st.median(df["Selling_Price"]),"\n")

print("----------Population Variance--------\n")
print(st.pvariance(df["Selling_Price"]),"\n")

print("----------Population Standard deviation--------\n")
print(st.pstdev(df["Selling_Price"]),"\n")

print("----------Mode--------\n")
print(st.mode(df["Selling_Price"]),"\n")

print("---------Correlation---------\n")
correlation1 = df['Kms_Driven'].corr(df["Selling_Price"])
correlation2 = df["Present_Price"].corr(df["Selling_Price"])
print("Correlation1:",correlation1)
print("Correlation2:",correlation2)
```

## 6. Pandas GroupBy Analysis

In this section, Pandas GroupBy is used to compare average selling prices across different car categories.

The following analysis is performed:

- Compare average selling prices by Fuel Type.
- Compare average selling prices by Transmission.
- Compare average selling prices by Seller Type.
- Analyze average selling prices by Fuel Type and Transmission together.

```python
## Pandas Groupby

print("---------Pandas Groupby-----------\n")

print("Average Selling Price by Fuel Type\n")
fuel_avg = df.groupby("Fuel_Type")["Selling_Price"].mean()
print(fuel_avg,"\n")

print("Average Selling Price by Transmission\n")
transmission_avg = df.groupby("Transmission")["Selling_Price"].mean()
print(transmission_avg,"\n")

print("Average Selling Price by Seller Type\n")
seller_avg = df.groupby("Seller_Type")["Selling_Price"].mean()
print(seller_avg,"\n")

print("Average Selling Price by Fuel Type and Transmission\n")
fuel_transmission_avg = df.groupby(["Fuel_Type","Transmission"])["Selling_Price"].mean()
print(fuel_transmission_avg,"\n")
```

## 8. Linear Algebra Analysis

In this section, Linear Algebra operations are performed using numerical data from the used car dataset.

The following operations are performed:

- **Matrix Addition:** Combine Dealer and Individual seller data based on fuel type. The matrices contain the number of cars and total selling price.
- **Matrix Multiplication:** Multiply a matrix by its inverse to obtain an Identity Matrix.
- **Determinant:** Calculate the determinant of a square matrix created from selected car price data.
- **Matrix Inverse:** Calculate the inverse of an invertible matrix.

These operations help understand basic Linear Algebra concepts using numerical data from the dataset.

```python
## Linear Algebra
## Matrix Addition
print("----------------------MATRIX ADDITION-----------------------------\n")
## Take the Unique Fuel_Type in the alphabetic order
print("---------Take the Unique Fuel_Type in the alphabetic order----------\n")
fuel_types = sorted(df["Fuel_Type"].unique())
print(fuel_types,"\n")

## dealer
print("------Filter by Dealer---------\n")
dealer = df[df['Seller_Type']=="Dealer"]
print(dealer,"\n")

## matrix A as for dealer
A = dealer.groupby("Fuel_Type").agg(
    car_count = ("Selling_Price","count"),
    total_price = ("Selling_Price","sum")
).reindex(fuel_types,fill_value = 0)

print("--------Before the  matrix----------\n")
print(A,"\n")

A = A.to_numpy()
print("--------After the  matrix----------\n")
print(A,"\n")

## individual
print("------Filter by Individual---------\n")
individual = df[df["Seller_Type"]=="Individual"]
print(individual,"\n")

## matrix B as for individual
B = individual.groupby("Fuel_Type").agg(
    car_count = ("Selling_Price","count"),
    total_price = ("Selling_Price","sum")
).reindex(fuel_types,fill_value = 0)

print("--------Before the  matrix----------\n")
print(B,"\n")

B = B.to_numpy()
print("--------After the  matrix----------\n")
print(B,"\n")

## Result in Matrix
print("-------Result in Matrix------------\n")
Result = A + B
print(Result,"\n")

## Answer in Dataframe
print("------------Answer in Dataframe----------\n")
print(pd.DataFrame(
    Result,
    index = fuel_types,
    columns = ["car_count","total_price"]
))

```

```python
## Matrix Multiplication
print("---------Matrix Multiplication-----------\n")
matrix = df[["Selling_Price","Present_Price"]].head(2).to_numpy()
inverse = np.linalg.inv(matrix)
product = matrix @ inverse
print(product.round(2))
```

```python
## Linear Algebra
## Determinant
print("----------Determinant value Finding---------------\n")
print("--------Convert as matrix--------------\n")

matrix = df[["Selling_Price","Present_Price"]].head(2).to_numpy()
print(matrix,"\n")

import numpy as np
det = np.linalg.det(matrix)
print("Determinant Value is:\n")
print(det)
print("------------------------------------------------------------------------------------------------------------\n")

## Matrix Inverse
print("---------Matrix Inverse-----------\n")
print(matrix,"\n")

inverse = np.linalg.inv(matrix)
print("The inverse matrix is:\n")
print(inverse)
```

## 11. Statistical Price Estimation

In this section, statistical methods are used to estimate the average selling price of a car based on its fuel type and transmission.

The following operations are performed:

- Group cars based on Fuel Type and Transmission.
- Calculate the average selling price for each group.
- Estimate the price of a Diesel car with Manual transmission using the corresponding group average.

This method provides an approximate price based on historical data.

**Note:** This is a statistical estimate using group averages, not a Machine Learning prediction model. The actual selling price may vary depending on the car's condition, age, and other factors.

```python
## Statistical Prediction

## Statistical Price Estimation

fuel_transmission_avg = df.groupby(["Fuel_Type","Transmission"])["Selling_Price"].mean()
print(fuel_transmission_avg,"\n")

fuel = "Diesel"
tran = "Manual"

detail = fuel_transmission_avg.loc[fuel,tran]
print("The estimated price is:",detail)
```
